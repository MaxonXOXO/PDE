# Topic 1: Quantum Confinement (The Schr&ouml;dinger Wave Equation)

## Phase 1: Theoretical Focus
The behavior of electrons in nanoscale electronics is governed by the 1D Time-Dependent Schr&ouml;dinger Equation:

$$i\hbar\frac{\partial\Psi(x,t)}{\partial t} = -\frac{\hbar^2}{2m}\frac{\partial^2\Psi(x,t)}{\partial x^2}$$

### Analytical Solution
For a particle in a 1D infinite potential well of length $L$ with boundary conditions $\Psi(0,t) = \Psi(L,t) = 0$, the normalized spatial part of the wave function is:

$$\psi_n(x) = \sqrt{\frac{2}{L}} \sin\left(\frac{n\pi x}{L}\right)$$

And the time-dependent part is:

$$\phi_n(t) = e^{-iE_n t / \hbar}$$

Where the quantized energy states are:

$$E_n = \frac{n^2\pi^2\hbar^2}{2mL^2}$$

## Phase 2: Computational Focus
We will now compute and plot the probability density function $|\Psi(x,t)|^2$ for the first three energy levels ($n=1, 2, 3$).

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
L = 1.0  # Length of the well
x = np.linspace(0, L, 1000)

def psi_n(x, n, L):
    """Spatial part of the wave function for a 1D infinite well"""
    return np.sqrt(2/L) * np.sin(n * np.pi * x / L)

# Plotting Wave Functions and Probability Densities
plt.figure(figsize=(12, 8))

for n in range(1, 4):
    psi = psi_n(x, n, L)
    prob_density = psi**2

    # Plot Wave Function
    plt.subplot(2, 1, 1)
    plt.plot(x, psi, label=f'n={n}')
    plt.title(r'Standing Waves $\psi_n(x)$')
    plt.xlabel('x')
    plt.ylabel(r'$\psi(x)$')
    plt.legend()

    # Plot Probability Density
    plt.subplot(2, 1, 2)
    plt.plot(x, prob_density, label=f'n={n}')
    plt.title(r'Probability Density $|\psi_n(x)|^2$')
    plt.xlabel('x')
    plt.ylabel(r'$|\psi(x)|^2$')
    plt.legend()

plt.tight_layout()
plt.show()
```

### Observations and Results
- The spatial wave function $\psi_n(x)$ represents the amplitude of the electron's position, showing $n-1$ nodes within the well.
- The probability density $|\psi_n(x)|^2$ shows where the electron is most likely to be found. For $n=1$, the peak is in the center, whereas for higher $n$, the probability distribution becomes more complex.
- This computational model validates the analytical solution derived from the 1D Schr&ouml;dinger equation.
